```python
import pandas as pd
import ast
import re

def parse_topics(x):
    if pd.isna(x):
        return []

    # already correct type
    if isinstance(x, list):
        return x

    if not isinstance(x, str):
        return []

    x = x.strip()

    # Case 1: proper Python list string
    if x.startswith('[') and x.endswith(']'):
        try:
            return ast.literal_eval(x)
        except:
            pass  # fallback below

    # Case 2: comma-separated
    if ',' in x:
        return [i.strip() for i in x.split(',') if i.strip()]

    # Case 3: pipe-separated (just in case)
    if '|' in x:
        return [i.strip() for i in x.split('|') if i.strip()]

    # Case 4: single value fallback
    return [x]

question_master['topics'] = question_master['topics'].apply(parse_topics)
```

```python
import ast
import pandas as pd
import re

def clean_tag(tag):

    if tag is None:
        return None

    tag = str(tag).strip()

    # remove common garbage
    tag = tag.strip("()[]{}")
    tag = tag.strip("'\"")
    tag = tag.strip(",")

    # normalize whitespace
    tag = re.sub(r"\s+", " ", tag)

    # canonical casing
    tag = tag.title()

    if not tag:
        return None

    return tag


def parse_topics(topic):

    if topic is None:
        return []

    if isinstance(topic, float) and pd.isna(topic):
        return []

    if isinstance(topic, (list, tuple, set)):
        return [
            clean_tag(t)
            for t in topic
            if clean_tag(t)
        ]

    if isinstance(topic, str):

        try:
            parsed = ast.literal_eval(topic)

            if isinstance(parsed, (list, tuple)):
                return [
                    clean_tag(t)
                    for t in parsed
                    if clean_tag(t)
                ]

            return [clean_tag(parsed)]

        except:
            pass

        # fallback for malformed strings
        topic = re.sub(r"[\(\)\[\]]", "", topic)

        return [
            clean_tag(t)
            for t in topic.split(",")
            if clean_tag(t)
        ]

    return []
```

```python
from collections import defaultdict

def get_company_tags(company_question, question_master):

    q_map = question_master.set_index('question_id')['topics'].to_dict()

    all_company_tags = {}

    for comp in company_question['company'].unique():

        company_tags = defaultdict(int)

        questions = company_question.loc[
            company_question['company'] == comp,
            'question_id'
        ]

        for q_id in questions:

            raw_topics = q_map.get(q_id)

            tags = parse_topics(raw_topics)

            for tag in tags:
                company_tags[tag] += 1

        all_company_tags[comp] = dict(company_tags)

    return all_company_tags

```

### COMPANY FEATURES ENGINEERING

```python
def get_company_features(company_question, question_master):

    q_map = question_master.set_index('ID')

    companies = company_question['company'].unique()

    results = []

    for comp in companies:

        comp_df = company_question[company_question['company'] == comp]

        total_questions = len(comp_df)
        easy = medium = hard = 0
        category_counts = {}

        acceptance = []
        frequencies = []
        likes = []
        dislikes = []

        for q_id in comp_df['ID']:

            if q_id not in q_map.index:
                continue

            row = q_map.loc[q_id]

            diff = row.get('difficulty_score')
            if diff == 1:
                easy += 1
            elif diff == 2:
                medium += 1
            elif diff == 3:
                hard += 1

            cat = row.get('category')
            if pd.notna(cat):
                category_counts[cat] = category_counts.get(cat, 0) + 1

            acc = row.get('acceptance_pct')
            if pd.notna(acc):
                acceptance.append(acc)

            freq = row.get('frequency_pct')
            if pd.notna(freq):
                frequencies.append(freq)

            likes_val = row.get('likes')
            if pd.notna(likes_val):
                likes.append(likes_val)

            dislikes_val = row.get('dislikes')
            if pd.notna(dislikes_val):
                dislikes.append(dislikes_val)

            comp_row = comp_df[comp_df['ID'] == q_id]
        

        results.append({
            'company': comp,
            'company_tier': company_question.loc[company_question['company'] == comp,'company_tier'].iloc[0],
            'total_questions': total_questions,

            'easy_pct': easy * 100 / total_questions if total_questions else 0,
            'medium_pct': medium * 100 / total_questions if total_questions else 0,
            'hard_pct': hard * 100 / total_questions if total_questions else 0,

            'avg_acceptance': pd.Series(acceptance).mean() if acceptance else 0,
            'median_acceptance': pd.Series(acceptance).median() if acceptance else 0,

            'avg_frequency': pd.Series(frequencies).mean() if frequencies else 0,
            'median_frequency': pd.Series(frequencies).median() if frequencies else 0,

            'avg_likes': pd.Series(likes).mean() if likes else 0,
            'avg_dislikes': pd.Series(dislikes).mean() if dislikes else 0,

            'top_category': max(category_counts, key=category_counts.get) if category_counts else None
        })

    return pd.DataFrame(results)
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertias = []

K = range(2,16)

for k in K:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init="auto"
    )

    model.fit(X_scaled)

    inertias.append(model.inertia_)

plt.figure(figsize=(7,5))

plt.plot(K, inertias, marker='o')

plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")

plt.title("Elbow Method")

plt.grid(True)

plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

scores = []

for k in range(2, 16):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init="auto"
    )

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)

    scores.append(score)

    print(f"k={k}: {score:.4f}")
```

```python
import matplotlib.pyplot as plt

plt.plot(range(2,16), scores, marker='o')
plt.xlabel("k")
plt.ylabel("Silhouette Score")
plt.grid(True)
plt.show()
```
